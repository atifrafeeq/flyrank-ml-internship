Generic accuracy is not useful here — the dataset has 54% declining pages, so a model that says "everything is declining" would score 54% accuracy and be useless. ## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*A fixed rule like "flag all pages older than 180 days with declining traffic" misses the interaction between signals. A page can be old but still growing. A page can be declining but have so few impressions it does not matter. A page can have low CTR because of its position, not its content quality. ML learns which combination of signals — age, impressions, sessions, trend, CTR, engagement — together predict a page worth reviewing. The starter pipeline already showed this: the random forest reached Precision@50 of 0.740 versus the baseline rule's 0.240. The pattern is real but too multi-dimensional for a single if-statement to capture.
