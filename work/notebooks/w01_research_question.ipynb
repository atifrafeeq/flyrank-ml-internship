## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import subprocess
import pandas as pd

# Clone repo if in Colab
if 'google.colab' in str(get_ipython()):
    if not os.path.exists('flyrank-ml-internship'):
        subprocess.run(['git', 'clone',
            'https://github.com/atifrafeeq/flyrank-ml-internship.git'])
    os.chdir('flyrank-ml-internship')

# Load starter dataset
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

# Real numbers to support Lane 2
print("=== Dataset Overview ===")
print(f"Total pages in dataset: {len(df)}")

# Number of declining pages
declining = df[df['trend_direction'] == 'down'].shape[0]
print(f"\nDeclining pages (trend = down): {declining}")
print(f"Percentage declining: {round(declining/len(df)*100, 1)}%")

# Stale pages (not updated in 180+ days)
stale = df[df['content_age_days'] >= 180].shape[0]
print(f"\nStale pages (180+ days old): {stale}")
print(f"Percentage stale: {round(stale/len(df)*100, 1)}%")

# High impression but declining pages
high_value_declining = df[
    (df['trend_direction'] == 'down') &
    (df['impressions_90d'] >= 500)
].shape[0]
print(f"\nHigh-impression declining pages (500+ impressions): {high_value_declining}")
print(f"These are the highest priority pages to review first.")
```
